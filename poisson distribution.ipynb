Suppose that $X$ has a Poisson distribution with parameter $\lambda = 3.2$.
This code uses the R function for the cdf of a Random Variable with Poisson distribution to determine, via trial and error, 
the smallest value of $x$ so that $P(X \leq x)$ first becomes greater than $0.9$.

```r
x = 0
p = 0
while (p<=0.9)
    
{
  p = ppois(x, 3.2, lower.tail = TRUE, log.p = FALSE)
  x = x + 1  
} 
print(x)
```

A plot showig that p reaches 0.9 at x=7

```r
xv = numeric(0)
x = 0
pv = numeric(0)
p = 0
while (p<=0.9)   
{
  p = ppois(x, 3.2, lower.tail = TRUE, log.p = FALSE)
  x = x + 1
  pv<-c(pv, p) 
  xv<-c(xv, x)
} 
print(x)
plot(xv, pv, xlim=c(0, 9), ylim=c(0, 1))
axis(1, at=seq(0, 9, by=1))
axis(2, at=seq(0, 1, by=0.1))
abline(v=x)
```
