This notebook contains my R code for calculating the probability of two or more students sharing a birthday, in a classroom of N students. This is the classic Birthday Problem in Probability. 
I calculated the probability on paper first, to check if my code generates a matching result. 
Let A be the event that at least two students share a birthday. There are 365 unique birthdays, so in a classroom of N students, the cardinality of the sample space is |S| = 365^N. Now calculating the cardinality of event A is quite tedious, and it's much easier to calculate the probability of the complement event Ac. Ac is the event that none of the N students share a birthday. In this case, the cardinality is an ordered sequence of N birthdays taken from the 365 days, i.e. it's a permutation of size N. 
Therefore |Ac| = 365!/(365 -N)! so P(A) = 1 - P(Ac) ==> P(A) = 1 - (365!/(365 -N)!)/365^N

Below is the R code I wrote to calculate P(A) using a for loop. I picked N = 40 haphazardly. 

```r
N<-40
#the probability for the first student not sharing a birthday with anyone 365/365 i = 0
probAc <- 1 
#using N - 1 instead of N because N is always ahead of i by one, because it includes the next person.
for (i in 0:(N-1)) {
#example: when i = 1 -> N = 2 because we're accounting for the 2nd students
#calculating the probability for each iteration, whereas on paper, we calculate them at once
    probAc <- probAc * (365 - i) / 365
     #i = 0 ==> prob=(365/365), i = 1 ==> prob=(365/365)*(364/365)...etc                                   
}
probA <- 1 - probAc
print(probA)
```

The probability of two or more students sharing a birthday P(A) increases with N. We can verify this by simply calculating it for different values of N. Moreover, looking back at the probablity of none of the students sharing the birthday P(Ac), we can see the opposite happening. At N = 1, P(Ac)=365/365=1, At N = 2, P(Ac)=(365/365)*(364/365)=0.99...
I wrote a code that finds the value of N required to reach a certain probability P(A). 
Let's say we want to find out at which N P(A)>0.8!
Here N becomes the unkown variable, so I used a while loop to calculate it. 

```r
N<-0
i<-0
probA<-0
probAc <- 1
while (probA<=0.8) 
{
    probAc <- probAc * (365 - (i+1)) / 365 
    i<- i + 1
    probA <- 1 - probAc #calculating P(A) inside the while loop
}
N = i + 1
print(N)
print(probA)
#the answer is N=35
```
