## Weak Law of Large Numbers

The **Weak Law of Large Numbers (WLLN)** states that if
\(X_1, X_2, \dots\) are independent and identically distributed
random variables with finite mean \(\mu\), then the sample mean

$$
\bar{X}_n = \frac{1}{n}\sum_{i=1}^{n}X_i
$$

**converges in probability to the population mean \(\mu\)** as
\(n \to \infty\):

$$
\boxed{\bar{X}_n \xrightarrow{P} \mu}
$$

In other words, as the sample size gets larger, the probability
that the sample mean is far from the true mean approaches zero.
