## The Central Limit Theorem (CLT)

Suppose we have a random/iid sample $X_1, X_2, \ldots, X_n$ from some distribution with mean $\mu$ and finite variance $\sigma^2 < \infty$.

Then the sample mean is

$$
\bar{X}_n = \frac{1}{n}\sum_{i=1}^n X_i
$$

The CLT states that as $n$ approaches infinity, the sample mean has an approximately normal distribution:

$$
\bar{X}_n \approx N\left(\mu, \frac{\sigma^2}{n}\right)
$$

or, equivalently,

$$
\frac{\bar{X}_n-\mu}{\sigma/\sqrt{n}}
\xrightarrow{d} N(0,1)
$$

Let's look at the uniform distribution $U(0,100)$ as an example.

First, let's write a function that computes the sample mean of a given sample of size $n$ from this uniform distribution.

```r
uniform.sample.mean = function(n){
    
    sample.mean = 0
    mysample = runif(n, 0, 100)
    for (i in 1:length(mysample))
        {
        sample.mean = sample.mean + mysample[i]/n
    }
    return(sample.mean)
}
```

Now let's check the normality of the sample mean distribution as $n$ gets large.

Let's simulate $m=2000$ sample means using the function we just wrote with a sample size of $n=10$.

Plotting a histogram of these sample means will show an approximately normal distribution.

```r
m = 2000
means = rep(0, m)
for (i in 1:m)
    {
    mean = uniform.sample.mean(10)
    means[i]= mean
    }
    #return(means) #breaks=seq(0,10,1)
hist(means,prob=T, breaks=30)
```
